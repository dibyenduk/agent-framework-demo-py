## 5. Basic Workflow - Executors and Edges

**Goal**: Build a simple Agent Framework workflow that transforms text using two executors.

```python
# Setup
from agent_framework import WorkflowBuilder, WorkflowContext, executor
from typing_extensions import Never
```

---
### Step 1 - Define the executors

- `upper_case` converts the input to uppercase and uses `send_message()` to pass it along an edge.
- `reverse_text` reverses the received text and uses `yield_output()` to return a workflow output.

```python
@executor(id="upper_case")
async def upper_case(text: str, ctx: WorkflowContext[str]) -> None:
    await ctx.send_message(text.upper())


@executor(id="reverse_text")
async def reverse_text(text: str, ctx: WorkflowContext[Never, str]) -> None:
    await ctx.yield_output(text[::-1])
```

---
### Step 2 - Connect the executors

`upper_case` is the starting executor. The edge routes its message to `reverse_text`, which is explicitly designated as the workflow's output executor.

```python
workflow = (
    WorkflowBuilder(start_executor=upper_case, output_from=[reverse_text])
    .add_edge(upper_case, reverse_text)
    .build()
)
```

---
### Step 3 - Run the workflow

```python
input_text = "hello world"
result = await workflow.run(input_text)
print(result.get_outputs()[0])
```
